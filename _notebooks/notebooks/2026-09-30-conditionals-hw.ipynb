## Popcorn Hack 2

```python

risk_score = 9

# Order matters: these checks go from the HIGHEST threshold to the LOWEST.
# If the order were reversed (checking >= 3 first, then >= 6, then >= 9),
# the >= 6 and >= 9 branches would become unreachable - any score of 3 or
# higher would always match the first, widest condition and stop there,
# so the more specific high-risk messages would never get printed.
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

## Homework Hack

```python
# CODE_RUNNER: 3.6 Homework - score a message, then advise on it

def score_message(message, amount_requested, known_contact):
    risk_score = 0

    # Signal 1: does the message mention gift cards?
    if "gift card" in message.lower():
        risk_score += 4
    # No else needed: not mentioning gift cards just means this signal adds
    # nothing - risk_score already starts at 0, so there is no action to undo.

    # Signal 2: is the amount requested unusually large?
    if amount_requested > 100:
        risk_score += 3
    # No else needed: a normal/small amount isn't suspicious on its own, so
    # "add nothing" is already the correct response - there's no penalty to remove.

    # Signal 3: is the sender NOT a known contact?
    if not known_contact:
        risk_score += 2
    # No else needed: a known contact simply means this signal doesn't fire -
    # there's no separate action required when the sender is trusted.

    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        advice = "Do not reply. This is almost certainly a scam."
    elif risk_score >= 6:
        advice = "This looks risky. Check with someone you trust before replying."
    elif risk_score >= 3:
        advice = "Be a little careful, but this is probably fine."
    else:
        advice = "Nothing suspicious in this message."

    print("Risk score:", risk_score)
    print(advice)
    print()
    return risk_score


# Given example
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
score_message(message, amount_requested, known_contact)

# Test the extremes: a completely clean message
score_message("Hey, are we still on for lunch tomorrow?", 20, True)

# Test the extremes: a message that trips every signal
score_message("URGENT gift card request, send $1000 immediately", 1000, False)
```

**Reachability proof** — one input set that reaches each answer:

| Inputs (message, amount, known_contact) | Score | Branch reached |
| --- | --- | --- |
| "gift card offer", 200, True | 7 | `>= 6`: "This looks risky..." |
| "quick meeting reminder", 150, True | 3 | `>= 3`: "Be a little careful..." |
| "URGENT: send gift cards now...", 500, False (the given example) | 9 | `>= 9`: "Do not reply..." |
| "Hey, are we still on for lunch tomorrow?", 20, True | 0 | `else`: "Nothing suspicious..." |

All four branches are reachable because the thresholds (9, 6, 3) are checked from highest to lowest — no branch sits behind a wider one that would catch its inputs first.

Stretch: Each scoring `if` here has no `else` because the default (add nothing) is already the correct outcome when the signal doesn't fire, whereas Example A's bug was leaving a variable with no value at all in the unhandled case — there, skipping the `else` meant something needed was missing, not that nothing needed to happen.
