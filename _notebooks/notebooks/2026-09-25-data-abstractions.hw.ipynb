## Popcorn

````python
# Popcorn Hack: Collections - Python vs AP CSP list indexing

# Original collection
levels = [12, 18, 10]  # star counts for Level 1, Level 2, Level 3 (Python index 0, 1, 2)

# Update an existing element (length stays the same)
levels[1] = 20  # Level 2's star count is updated

# Append a new element (length increases)
levels.append(5)  # a new Level 4 is added with 5 stars

print(levels)
print(len(levels))
````

**AP CSP pseudocode equivalent:**

```
levels ← [12, 18, 10]
levels[2] ← 20      // AP index 2 = Python index 1 = second level
APPEND(levels, 5)

DISPLAY(levels)
DISPLAY(LENGTH(levels))
```

````python
# Rerun: appended value changed from 5 to 8
levels = [12, 18, 10]
levels[1] = 20
levels.append(8)

print(levels)
print(len(levels))
````

Prediction: Updating `levels[1]` to 20 keeps the collection at 3 elements — `[12, 20, 10]` — since it only replaces the value already stored at an existing index. Appending 5 adds a brand-new element, growing the collection to `[12, 20, 10, 5]` with a level count of 4.

AP index: Python index 1 (the second level) matches AP CSP pseudocode index 2, since College Board's pseudocode lists start counting at 1 instead of 0.

Actual output (append 5): `[12, 20, 10, 5]`, length 4. Matches the prediction.

Actual output (append 8 instead of 5): `[12, 20, 10, 8]`, length 4 — same length as the append-5 version, since appending always adds exactly one new element no matter what value is appended.

Why updating keeps the length the same but appending increases it: `levels[1] = 20` overwrites the value already stored at an existing index, so no new slot is created and `len(levels)` doesn't change. `levels.append(x)` creates a brand-new slot at the end of the collection, so the length always grows by one regardless of the value being appended.

## Homework

````python
# Homework Hack: refactor the 3.1 prototype into a list

# Each element is one level's collected stars, in level order:
# level_stars[0] = Level 1, level_stars[1] = Level 2, level_stars[2] = Level 3
level_stars = [12, 18, 10]

# Add 5 to the second level
level_stars[1] = level_stars[1] + 5

# Append 7 for a new level
level_stars.append(7)

# Example C traversal: walk the collection to total the stars
total = 0
for stars in level_stars:
    total = total + stars

print(level_stars)
print(len(level_stars))
print(total)
````

Prediction: `level_stars` becomes `[12, 23, 10, 7]`, the level count is 4, and the total is 52.

Actual output: `[12, 23, 10, 7]`, `4`, `52`. Matches the prediction.

Compared to 100 separate variables (`level_1_stars`, `level_2_stars`, ..., `level_100_stars`), a list avoids: naming 100 unique variables, writing 100 separate update/print statements, and hardcoding which variable to touch whenever a level is added or removed. With a list, adding a new level is a single `.append()` call, and Example C's traversal (the `for` loop) totals or prints any number of levels without editing the code — the loop works the same whether there are 3 levels or 300.

## Tests

````python
# Test Hack: reusable update + traversal, guarded for a missing second level

def add_bonus_to_second_level(levels, bonus):
    if len(levels) > 1:  # skip the second-level update when there is no second element
        levels[1] = levels[1] + bonus
    return levels

def summarize(levels):
    total = 0
    for stars in levels:
        total = total + stars
    return len(levels), total

# Test the total-only code with an empty collection
test_levels = []
add_bonus_to_second_level(test_levels, 5)
count, total = summarize(test_levels)

print(count, total)
````

Prediction: With an empty collection, there is no second element, so the bonus update is skipped. The traversal then finds a level count of 0 and a total of 0.

Actual output: `0 0`. Matches the prediction.

Connection to 3.1: In the 3.1 homework, changing `level_two_stars` after `total_stars` was already calculated did not update `total_stars` on its own — recalculating it required rerunning the addition. The same idea holds here: editing `level_stars` (updating an element or appending one) does not retroactively recalculate an already-stored total. The list only organizes the state; it's the assignment (`level_stars[1] = ...`) and Example C's traversal (the `for` loop that rebuilds `total`) that actually update the values people see.

## Design Thinking

**Player/maker's need:** A player or level designer needs to track and adjust star counts across many levels without editing a wall of separate variables every time a level is added, removed, or rebalanced.

**Goal:** Represent all levels' star counts in one collection that can grow, shrink, and be totaled with the same few lines of code no matter how many levels exist.

**Representation considered:** I considered keeping `level_one_stars`, `level_two_stars`, `level_three_stars`, etc. as separate variables (like in 3.1), but that doesn't scale — every new level needs a new variable name and new print/update lines. A list (`level_stars`) lets every level share one name and be reached by index or by looping.

**Prototype built:** `level_stars = [12, 18, 10]`, updated in place (index 1 `+= 5`), grown with `.append(7)`, and summarized with Example C's traversal (a `for` loop) to print the collection, its length, and its total.

**Test/revision:** I tested the update-and-total logic against an empty list (`[]`) and found it would crash trying to update `levels[1]` with no second element — so I added a length guard (`if len(levels) > 1`) before the update, which fixed the crash and correctly produced `0, 0` for the empty case.

**Suggestion — clear level labels:** Right now, editing this list means knowing that index 1 means "the second level," which isn't obvious. Pairing the list with clear labels (e.g., a parallel list of names like `["Forest", "Cave", "Castle"]`, or short comments naming each index) would let players or other developers understand which level they're changing without needing to think in terms of raw indices.
