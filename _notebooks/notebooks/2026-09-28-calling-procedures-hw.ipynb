```python
#Popcorn Hack 1
def report_status(location, status):
    print("Status Update from " + location + ": " + status)


report_status("Poway", "Clear")
report_status("Downtown", "Under Investigation")
```

```python
#Popcorn Hack 2
def is_priority(incident_type):
    return incident_type == "Medical Emergency"


def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print(f"PRIORITY: {incident_type} reported at {location}")
    else:
        print(f"Report: {incident_type} at {location}")

process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

```python
# Homework Hack
def check_priority(incident_type):
    return incident_type in ("Medical Emergency", "Fire", "Crime in Progress")

def dispatch_response(location, incident_type):
    is_urgent = check_priority(incident_type)

    if is_urgent:
        print(f"URGENT: Dispatching units to {location} for {incident_type}")
    else:
        print(f"Logged: {incident_type} at {location} will be reviewed later")

dispatch_response("Poway", "Medical Emergency")
dispatch_response("Downtown", "Noise Complaint")
dispatch_response("Escondido", "Fire")
```
