## Design Thinking

**Player's need:** Some players are sensitive to fast motion, flashing, or screen shake and can feel discomfort (dizziness, headaches, or eye strain) from effects that other players enjoy.

**My goal:** Let players opt into a calmer visual experience without losing any game functionality, like star totals or progress tracking.

**Setting/representation I considered:** I considered a string setting like `"on"`/`"off"`, but chose a Boolean (`reduced_motion`) instead, since the setting is strictly one of two states and a Boolean is simpler to check in an if-statement.

**Prototype I built:** Variables tracking a game name, a `reduced_motion` Boolean, and per-level star counts, with a total that recalculates correctly after a bonus is applied to one level.

**Test/revision:** I tested the bonus logic twice — once with `bonus = 5` (stored total 40, updated total 45) and once with `bonus = 0` (stored total 40, updated total still 40) — confirming the total only changes when a real bonus is added, and the recalculation logic isn't accidentally changing values on its own.

**Why reduced motion should be a player choice:** Motion sensitivity varies from person to person, and a single fixed visual style can't work for everyone. Making `reduced_motion` a setting the player controls, rather than a fixed design decision, keeps the game accessible to players who need it while still letting other players enjoy full motion effects.
