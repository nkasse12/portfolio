## Homework

```python
# Example C adapted for the library catalog

def example_c(n):
    visits = 0
    ordered_pairs = 0
    for i in range(n):
        visits += 1
        for j in range(n):
            ordered_pairs += 1
    return n, visits, ordered_pairs

book_count, visits, ordered_pairs = example_c(12)

print("Book count:", book_count)
print("Visits:", visits)
print("Ordered pairs:", ordered_pairs)
```

Prediction: With n = 12 books, I expect the book count to print 12, visits to print 12 (one visit per book), and ordered pairs to print 144 (12 × 12, since every book is paired with every book).

Actual output: Book count: 12, Visits: 12, Ordered pairs: 144. Matches the prediction.

Explanation: `visits` grows with the single outer loop, so it scales linearly with `n`. `ordered_pairs` grows with the nested loop (an inner loop of size `n` running once per outer visit), so it scales with `n * n` — quadratically.

```python
# Double the catalog to n = 24

book_count, visits, ordered_pairs = example_c(24)

print("Book count:", book_count)
print("Visits:", visits)
print("Ordered pairs:", ordered_pairs)
```

Prediction: Doubling the catalog to n = 24, I expect book count 24, visits 24, and ordered pairs 576 (24 × 24).

Actual output: Book count: 24, Visits: 24, Ordered pairs: 576. Matches the prediction.

Different growth rates: doubling `n` from 12 to 24 exactly **doubled** `visits` (12 → 24), since it only depends on the single outer loop — that's linear growth. But `ordered_pairs` went from 144 to 576, which is **4×**, not 2× — because the nested loop means ordered pairs scale with `n * n`: doubling `n` always quadruples `n²` (since (2n)² = 4n²). That's the difference between linear and quadratic growth.

```python
# Checkpoint B: worst-case checks for 1,000 books

n = 1000
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print("Catalog size:", n)
print("Linear worst-case checks:", linear_checks)
print("Binary worst-case checks:", binary_checks)
```

Recommendation: for 1,000 books, linear worst-case checks is 1,000 versus binary's 10 — binary search is the clear choice for **repeated** searches of a catalog that's already sorted, since every search afterward costs at most 10 checks instead of up to 1,000.

What changes if the catalog is unsorted: binary search requires sorted data to work at all, so an unsorted catalog would need to be sorted first — that sort itself costs more than a single linear search, so for a *one-time* search, sorting just to binary-search isn't worth it. It only pays off once you're doing many repeated searches.

What changes if you must store an extra sorted copy: keeping a separate sorted copy avoids re-sorting the original each time, but costs extra memory (roughly double the storage for the catalog) and extra work to keep the sorted copy in sync whenever a book is added or removed.

## Design Thinking

**Librarian's need:** A librarian needs to know, without reading any code, roughly how much work a search method will take as the catalog grows — so they can decide whether it's worth sorting the shelves for faster lookups.

**Goal:** Compare how two approaches (checking every book one by one vs. jumping through a sorted catalog) scale as the number of books grows, and express that comparison in numbers a non-programmer can read directly.

**Methods considered:** I considered reporting raw timing (seconds), but that depends on the computer running it and isn't a fair comparison. I used **check counts** instead (how many books get looked at), since that's a property of the method itself, independent of hardware.

**Prototype built:** `example_c(n)` reports book count, visits, and ordered pairs for the nested-loop model (n and n² growth), and the checkpoint B block reports linear vs. binary worst-case checks for a 1,000-book catalog.

**Test/revision:** I tested `example_c` with an empty catalog (n = 0) and confirmed it returns all zeros with no error, rather than crashing — an edge case worth checking since `range(0)` could easily have been mishandled.

**Clear output labels:** Every print statement uses a plain-English label ("Book count", "Visits", "Ordered pairs", "Linear worst-case checks", "Binary worst-case checks") instead of the bare variable names — so a librarian reading the output doesn't need to know what `n`, `visits`, or `ordered_pairs` mean in the formulas to understand the result.

**Connection to the warm-up:** A method with better worst-case growth (binary search) can still take more checks than linear search on a *particular* target — for example, if the very first linear check happens to land on the right book, linear search wins that one case even though it loses badly overall. This is why it's important to separate three different ideas: **correctness** (does it find the right answer at all), **individual cases** (how many checks a specific search happens to take), and **growth** (how the worst case scales as the catalog gets bigger). Binary search's advantage is about growth, not about guaranteeing the fewest checks on every single search.
